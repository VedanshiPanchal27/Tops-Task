```python
#task 1 

def bayes_posterior(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior


prior = 0.10
likelihood = 0.80
evidence = 0.26

result = bayes_posterior(prior, likelihood, evidence)

print("Posterior probability:", result)
```

```python
#task 2 

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


prior_premium = 0.10

likelihood_discount = 0.80

prior_regular = 1 - prior_premium

likelihood_discount_regular = 0.20

evidence = (
    likelihood_discount * prior_premium
    + likelihood_discount_regular * prior_regular
)

posterior = bayes_posterior(
    prior_premium,
    likelihood_discount,
    evidence
)

print("Probability restaurant is premium:", posterior)
print("Percentage:", posterior * 100, "%")
```

```python
#task 3 

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence


prior_spam = 0.30

likelihood_free_spam = 0.70

prior_not_spam = 1 - prior_spam

likelihood_free_not_spam = 0.10

evidence = (
    likelihood_free_spam * prior_spam
    + likelihood_free_not_spam * prior_not_spam
)

posterior = bayes_posterior(
    prior_spam,
    likelihood_free_spam,
    evidence
)

print("Probability email is spam:", posterior)
print("Percentage:", posterior * 100, "%")
```

```python
#task 4 

def bayes_posterior(prior, likelihood, evidence):
    return (prior * likelihood) / evidence

prior_phone = 0.20

likelihood_case_phone = 0.60

prior_no_phone = 1 - prior_phone

likelihood_case_no_phone = 0.10

evidence = (
    likelihood_case_phone * prior_phone
    + likelihood_case_no_phone * prior_no_phone
)

posterior = bayes_posterior(
    prior_phone,
    likelihood_case_phone,
    evidence
)

print("Probability user bought a phone given they bought a case:",
      posterior)

print("Percentage:", posterior * 100, "%")
```
