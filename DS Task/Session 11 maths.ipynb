```python
#task 3 

orders = [
    [10, 20, 15, 12],
    [12, 18, 20, 10],
    [15, 22, 18, 14]
]

for dish in range(4):
    total = 0

    for day in range(3):
        total += orders[day][dish]

    print("Dish", dish + 1, "total orders:", total)
```

```python
#task 4 

products = [
    [500, 10],
    [1000, 5],
    [250, 20],
    [750, 8],
    [1200, 4]
]


def calculate_revenue(products):
    for product in products:
        price = product[0]
        quantity = product[1]

        revenue = price * quantity

        print("Revenue:", revenue)


calculate_revenue(products)
```
