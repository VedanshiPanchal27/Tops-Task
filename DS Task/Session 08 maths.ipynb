```python
#task 1 

def calculate_probability(event_count, total_outcomes):
    return event_count / total_outcomes

heads = 1
total_outcomes = 2

probability = calculate_probability(heads, total_outcomes)

print("Probability of Heads:", probability)
```

```python
#task 2 

import random 

count_six = 0

for i in range(100):
    dais = random.randint(1,6)

    if dais == 6:
        count_six+=1

probability = count_six/100

print("probability:-",probability)
```

```python
#task 3 

genres = [
    'pop', 'rock', 'pop', 'hiphop', 'pop',
    'jazz', 'rock', 'pop', 'hiphop', 'pop',
    'rock', 'jazz', 'pop', 'pop', 'rock',
    'hiphop', 'jazz', 'pop', 'rock', 'pop'
]

pop_count = genres.count('pop')

total_songs = len(genres)

probability = pop_count / total_songs

print("Number of Pop songs:", pop_count)
print("Total songs:", total_songs)
print("Probability of picking Pop:", probability)
```

```python
#task 4

def probability_both(food_users, dessert_users):
    food_users = set(food_users)
    dessert_users = set(dessert_users)

    both_users = food_users.intersection(dessert_users)

    all_users = food_users.union(dessert_users)

    probability = len(both_users) / len(all_users)

    return probability


food_users = ['user1', 'user2', 'user3', 'user4', 'user5']
dessert_users = ['user3', 'user4', 'user6', 'user7']

result = probability_both(food_users, dessert_users)

print("Probability:", result)
```

```python
#task 5

def conditional_probability(both, phone_users):
    return both / phone_users


phone_users = 100
both = 30

probability = conditional_probability(both, phone_users)

print("Probability of buying headphones given phone:", probability)
print("Percentage:", probability * 100, "%")
```
