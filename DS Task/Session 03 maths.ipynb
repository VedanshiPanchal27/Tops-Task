```python
# 1.Given an array of daily step counts from your fitness app for one week (e.g., [4500, 7000, 5000, 8000, 4000, 9000, 6000]), calculate the variance by hand using the variance formula and show each calculation step.
# Step 1: Find the mean
# First, I add all the step counts and divide by the total number of days.
# 4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000 = 43500
# There are 7 days, so:
# Mean = 43500 / 7
#      = 6214.29
# Step 2: Find how far each value is from the mean
# For example:
# 4500 - 6214.29 = -1714.29
# 7000 - 6214.29 = 785.71
# We do this for all 7 values.
# Step 3: Square each difference
# We square the differences because negative and positive differences should not cancel each other.
# For example:
# (-1714.29)² = 2,938,775.51
# (785.71)²   = 617,346.94
# Step 4: Add all squared differences
# Total = 20,928,571.43
# Step 5: Divide by 7
# Variance = 20,928,571.43 / 7
#          = 2,989,795.92
```

```python
# 2.Write a Python function called calculate_standard_deviation(scores) that takes a list of exam scores and returns the standard deviation rounded to two decimal places.
def calculate_standard_deviation(scores):
    mean = sum(scores) / len(scores)

    squared_deviations = [(x - mean) ** 2 for x in scores]

    variance = sum(squared_deviations) / len(scores)

    standard_deviation = variance ** 0.5

    return round(standard_deviation, 2)


scores = [80, 85, 90, 75, 95]

print(calculate_standard_deviation(scores))
```
