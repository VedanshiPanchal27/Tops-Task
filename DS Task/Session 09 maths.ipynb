```python
#task1 

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 100)

print(data)

plt.hist(data, bins=10, edgecolor="black")

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution")

plt.show()
```

```python
#task 2 

import numpy as np
from scipy import stats

steps = [
    5000, 5200, 5100, 5300, 5000,
    5400, 5100, 5200, 5000, 5300,
    5100, 5200, 5400, 5000, 5100,
    5300, 5200, 5100, 5000, 5200
]

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=True)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode.mode[0])
```

```python
#task 3 

import numpy as np
scores = np.random.normal(70, 15, 200)
count = np.sum((scores >= 55) & (scores <= 85))
print("Players within one standard deviation:", count)
print("Percentage:", (count / 200) * 100)
```

Zomato delivery times

Suppose we collect delivery times from a large number of similar orders. Most orders might be delivered around the average time, while fewer orders are extremely fast or extremely slow.

If the data is approximately symmetrical around the average, a normal distribution can be useful for modeling it. It allows us to understand the average delivery time and how much delivery times vary around that average. This can help Zomato identify unusually fast or slow deliveries.
