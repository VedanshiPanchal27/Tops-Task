```python
# 1.Install Matplotlib in your Python environment and create a simple line plot showing the number of daily steps you took over the last 7 days. Save the chart as steps_lineplot.png.
!pip install matplotlib
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [6500, 8200, 7500, 9100, 8800, 10500, 7200]

plt.plot(days, steps, marker="o")

plt.title("Daily Steps - Last 7 Days")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

plt.grid(True)

plt.savefig("steps_lineplot.png")
plt.show()
```

```python
# 2.Using Matplotlib, create a scatter plot of 10 restaurants from your city with their Zomato ratings on the x-axis and average meal price on the y-axis. Add axis labels and a title to the chart.
import matplotlib.pyplot as plt

restaurants = [
    "Dominos", "Pizza Hut", "Honest", "La Pinoz", "McDonalds",
    "Subway", "Swiggy Cafe", "TGB", "Manek Chowk", "Urban Khichdi"
]

ratings = [4.2, 4.5, 3.9, 4.7, 4.1, 4.4, 3.8, 4.6, 4.0, 4.3]

meal_price = [350, 500, 250, 700, 300, 450, 220, 650, 280, 400]

plt.scatter(ratings, meal_price)

plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price (₹)")
plt.title("Zomato Ratings vs Average Meal Price")

# Add restaurant names
for i in range(len(restaurants)):
    plt.annotate(restaurants[i], (ratings[i], meal_price[i]))

plt.grid(True)
plt.show()
```

```python
# 3.Build a bar chart that shows the number of orders you or your friends placed on Swiggy, Zomato, and Domino’s in the last month. Use different colors for each bar and add a legend.
import matplotlib.pyplot as plt

platforms = ["Swiggy", "Zomato", "Domino's"]
orders = [18, 14, 9]

colors = ["orange", "red", "blue"]

plt.bar(platforms, orders, color=colors, label="Number of Orders")

plt.title("Food Orders in Last Month")
plt.xlabel("Food Delivery Platform")
plt.ylabel("Number of Orders")

plt.legend()

plt.show()
```

```python
# 4.Create a histogram of the durations (in minutes) of your last 20 Spotify listening sessions using Matplotlib. Set the number of bins to 5 and customize the color of the bars.
import matplotlib.pyplot as plt

durations = [
    25, 40, 35, 55, 30,
    45, 60, 20, 50, 38,
    42, 28, 65, 33, 48,
    52, 27, 70, 36, 44
]

plt.hist(
    durations,
    bins=5,
    color="mediumseagreen",
    edgecolor="black"
)

plt.title("Spotify Listening Session Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Sessions")

plt.show()
```

```python
# 5.Customize a Matplotlib figure by creating a plot with two subplots: one line plot showing your daily Instagram screen time for a week, and one bar chart showing the number of posts you liked each day. Add appropriate titles, axis labels, and save the figure as social_media_usage.png.<br><br><em><strong>Hint:</strong> Use plt.subplots() to create multiple axes in one figure.</em>
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

screen_time = [2.5, 3.0, 2.2, 4.0, 3.5, 5.0, 4.2]

liked_posts = [12, 18, 10, 25, 20, 30, 22]

# Create two subplots
fig, axes = plt.subplots(2, 1, figsize=(9, 8))

# First subplot - Line Plot
axes[0].plot(days, screen_time, marker="o")

axes[0].set_title("Daily Instagram Screen Time")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Screen Time (Hours)")

axes[0].grid(True)

# Second subplot - Bar Chart
axes[1].bar(days, liked_posts)

axes[1].set_title("Instagram Posts Liked Each Day")
axes[1].set_xlabel("Day")
axes[1].set_ylabel("Number of Posts Liked")

# Adjust layout
plt.tight_layout()

# Save the complete figure
plt.savefig("social_media_usage.png")

plt.show()
```
