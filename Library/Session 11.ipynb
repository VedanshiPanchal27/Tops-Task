```python
# 1.Use plt.subplots() to create a 2x2 grid of subplots and plot four different types of charts (line, bar, scatter, and pie) using any sample data of your choice.
import matplotlib.pyplot as plt

# Sample data
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
steps = [5000, 7000, 6500, 8000, 9000]

platforms = ["Swiggy", "Zomato", "Domino's"]
orders = [20, 15, 10]

ratings = [3.5, 4.0, 4.5, 5.0, 4.2]
prices = [200, 300, 450, 600, 400]

food_types = ["Pizza", "Burger", "Indian", "Chinese"]
orders_food = [30, 25, 20, 15]

# Create 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1. Line chart
axes[0, 0].plot(days, steps, marker="o")
axes[0, 0].set_title("Daily Steps")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Steps")

# 2. Bar chart
axes[0, 1].bar(platforms, orders)
axes[0, 1].set_title("Food Delivery Orders")
axes[0, 1].set_xlabel("Platform")
axes[0, 1].set_ylabel("Orders")

# 3. Scatter plot
axes[1, 0].scatter(ratings, prices)
axes[1, 0].set_title("Rating vs Meal Price")
axes[1, 0].set_xlabel("Rating")
axes[1, 0].set_ylabel("Price (₹)")

# 4. Pie chart
axes[1, 1].pie(
    orders_food,
    labels=food_types,
    autopct="%1.1f%%"
)
axes[1, 1].set_title("Food Preferences")

# Adjust spacing
plt.tight_layout()

plt.show()
```

```python
# 2.Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options.
import matplotlib.pyplot as plt

platforms = ["Zomato", "Swiggy", "Domino's"]
delivery_time = [35, 30, 25]

# Different styling for each bar
colors = ["red", "orange", "blue"]
linestyles = ["-", "--", ":"]
linewidths = [2, 3, 4]

fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(platforms, delivery_time)

# Style each bar
for bar, color, linestyle, linewidth in zip(
    bars, colors, linestyles, linewidths
):
    bar.set_facecolor(color)
    bar.set_edgecolor("black")
    bar.set_linestyle(linestyle)
    bar.set_linewidth(linewidth)

ax.set_title("Average Delivery Time")
ax.set_xlabel("Food Delivery Platform")
ax.set_ylabel("Average Delivery Time (Minutes)")

plt.show()
```

```python
# 3.Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.
import matplotlib.pyplot as plt

influencers = [
    "Influencer A",
    "Influencer B",
    "Influencer C",
    "Influencer D",
    "Influencer E"
]

followers = [120000, 250000, 180000, 350000, 500000]

daily_posts = [2, 4, 3, 5, 6]

fig, ax1 = plt.subplots(figsize=(10, 6))

# Left Y-axis - Followers
ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Instagram Followers", color="blue")

# Right Y-axis - Daily Posts
ax2 = ax1.twinx()
ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts", color="red")

plt.title("Instagram Followers vs Average Daily Posts")

plt.tight_layout()
plt.show()
```

```python
# 4.Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point.
import matplotlib.pyplot as plt

movies = [
    "Movie A",
    "Movie B",
    "Movie C",
    "Movie D",
    "Movie E"
]

tickets_sold = [2.5, 4.0, 3.2, 5.5, 4.5]  # Millions

imdb_rating = [7.2, 8.1, 6.8, 8.5, 7.8]

plt.figure(figsize=(9, 6))

plt.scatter(
    tickets_sold,
    imdb_rating,
    s=100
)

# Add movie names above each point
for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], imdb_rating[i]),
        xytext=(5, 8),
        textcoords="offset points"
    )

plt.title("Bollywood Movies: Tickets Sold vs IMDb Rating")
plt.xlabel("Tickets Sold (Millions)")
plt.ylabel("IMDb Rating")

plt.grid(True)

plt.show()
```

```python
# 5.Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>
import matplotlib.pyplot as plt

months = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

sales = [
    120, 150, 135, 160,
    175, 180, 165, 190,
    210, 350, 230, 200
]

fig, ax = plt.subplots(figsize=(10, 6))

# Line chart
ax.plot(
    months,
    sales,
    marker="o",
    linewidth=2
)

# Highest sales point
highest_sales = max(sales)
highest_index = sales.index(highest_sales)

# Add annotation
ax.annotate(
    "Big Billion Days",
    xy=(months[highest_index], highest_sales),
    xytext=(20, 30),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (₹ Lakhs)")

ax.grid(True)

plt.tight_layout()
plt.show()
```
